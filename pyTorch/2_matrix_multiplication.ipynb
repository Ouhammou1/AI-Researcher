```python
import numpy as np
import torch 


A = torch.tensor([[1,2],
                 [3, 4]])


B = torch.tensor([[5, 6],
                 [7, 8]])


print("MATRIX A")
print(A)
print(A.shape)

print()



print("MATRIX B")
print(B)
print(B.shape)
```

```python
C = A@B


print("MATRIX C")
print(C)
print(C.shape)

      #using matmul


res2 = torch.matmul(A,B)
print("MATMUL ")

print(res2)
```
