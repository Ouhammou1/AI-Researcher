```python
import numpy as np
import matplotlib.pyplot as plt

# définir la fonction
def f(x):
    return x**2

# domaine
x = np.linspace(-4, 4, 400)
y = f(x)

# deux points x1, x2
x1, x2 = -3, 2
y1, y2 = f(x1), f(x2)

# lambda
lam = 0.4

# point entre x1 et x2
x_mid = lam * x1 + (1 - lam) * x2
y_mid = f(x_mid)

# droite entre (x1,f(x1)) et (x2,f(x2))
line_x = np.linspace(x1, x2, 100)
line_y = lam * y1 + (1 - lam) * y2 + (line_x - x_mid) * 0  # droite horizontale visuelle

# tracé
plt.figure(figsize=(6,4))
plt.plot(x, y, label="f(x) = x²", linewidth=2)
plt.plot([x1, x2], [y1, y2], 'r--', label="Droite entre deux points")
plt.scatter([x1, x2], [y1, y2], color='red')
plt.scatter(x_mid, y_mid, color='green', label="Point intermédiaire")

plt.axhline(0, color='black', linewidth=0.5)
plt.axvline(0, color='black', linewidth=0.5)

plt.legend()
plt.xlabel("x")
plt.ylabel("f(x)")
plt.title("Convexité de f(x) = x²")
plt.grid(True)
plt.show()
```
