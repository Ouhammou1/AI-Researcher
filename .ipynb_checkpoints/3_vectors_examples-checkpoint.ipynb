```python
import numpy as np
import matplotlib.pyplot as plt
    
```

```python
v1 = np.array([3,4])
v2 = np.array([-2,1])
v3 = np.array([1,5])


fig ,ax= plt.subplots(figsize=(7,5))
ax.quiver(0, 0 ,v1[0] , v1[1] , angles='xy' , scale_units='xy' , scale=1 , color='r' , width=0.005  , label='v1 = {v1}')
ax.quiver(0, 0 ,v2[0] , v2[1] , angles='xy' , scale_units='xy' , scale=1 , color='b' , width=0.005  , label='v1 = {v1}')
ax.quiver(0, 0 ,v3[0] , v3[1] , angles='xy' , scale_units='xy' , scale=1 , color='g' , width=0.005  , label='v1 = {v1}')


ax.set_xlim(-5, 5)
ax.set_ylim(-1, 6)
# ax.set_xlabel('x')
# ax.set_ylabel('y')
# ax.set_title('Vectors as Arrows')
# ax.grid(True)
# ax.legend()
# ax.set_aspect('equal')
# plt.axhline(y=0, color='k', linewidth=0.5)
# plt.axvline(x=0, color='k', linewidth=0.5)

plt.grid()
plt.show()


```

```python
v1 = np.array([3,4])
v2 = np.array([-2,1])
v3 = np.array([1,5])


fig ,ax= plt.subplots(figsize=(7,5))
ax.quiver(0, 0 ,v1[0] , v1[1] , angles='xy' , scale_units='xy' , scale=1 , color='r' , width=0.005  , label='v1 = {v1}')
ax.quiver(v1[0] , v1[1]  ,v2[0] , v2[1] , angles='xy' , scale_units='xy' , scale=1 , color='b' , width=0.005  , label='v1 = {v1}')
ax.quiver(0, 0 ,v3[0] , v3[1] , angles='xy' , scale_units='xy' , scale=1 , color='g' , width=0.005  , label='v1 = {v1}')


ax.set_xlim(-5, 5)
ax.set_ylim(-1, 6)
ax.set_xlabel('x')
ax.set_ylabel('y')
ax.set_title('Vectors as Arrows')
ax.grid(True)
ax.legend()
ax.set_aspect('equal')
plt.axhline(y=0, color='k', linewidth=0.5)
plt.axvline(x=0, color='k', linewidth=0.5)

plt.grid()
plt.show()


```
