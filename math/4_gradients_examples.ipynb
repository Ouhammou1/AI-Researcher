```python
X = [1, 2, 3]
Y = [2, 4, 6]


w = 1.0

lr = 0.01

weight_history = []
loss_history = []

for epoch in range(24):
    total_loss = 0
    gradient_sum = 0

    for x , y in zip(X,Y):
        y_pred = w * x
        error = y - y_pred
        loss = error ** 2
        total_loss = total_loss + loss


        
        gradient = -2 * x * error
        gradient_sum = gradient_sum + gradient

    w = w - lr * gradient_sum
    print(f"Epoch {epoch+1}: w = {w:.3f}, y = {y:.3f}, y_pred = {y_pred:.3f},  error = {error:.3f}, loss = {total_loss:.3f}")



    weight_history.append(w)
    loss_history.append(loss)
    
    plt.figure(figsize=(7,5))
    plt.scatter(X, Y , color="blue" , label="Real points")
    pred_line = [w * x for  x in X]
    plt.plot(X , pred_line, color="red", label=f"Prediction (w={w:.3f})")
    plt.title(f"Epoch {epoch+1} — Prediction Line")
    plt.xlabel("X")
    plt.ylabel("Y")
    plt.legend()
    plt.grid(True)
    plt.show()



plt.figure(figsize=(6, 4))
plt.plot(weight_history, marker="o")
plt.title("Weight Evolution")
plt.xlabel("Epoch")
plt.ylabel("Weight (w)")
plt.grid(True)
plt.show()

# Loss evolution
plt.figure(figsize=(6, 4))
plt.plot(loss_history, marker="o")
plt.title("Loss Evolution")
plt.xlabel("Epoch")
plt.ylabel("Total Loss")
plt.grid(True)
plt.show()
```

```python
# Training data
X = [1, 2, 3]
Y = [2, 4, 6]   # real function y = 2x

# model weight (start badly)
w = 1.0

# learning rate
lr = 0.1

# Training loop
for epoch in range(100):
    total_loss = 0
    gradient_sum = 0
    
    # Compute predictions and gradient
    for x, y in zip(X, Y):
        y_pred = w * x                 # prediction
        error = y - y_pred             # error
        loss = error ** 2              # squared loss
        total_loss += loss
        
        # gradient of (y - w*x)^2 wrt w = -2*x*(y - w*x)
        gradient = -2 * x * error
        gradient_sum += gradient
    
    # Update weight
    w = w - lr * gradient_sum          # gradient descent step
    
    # print(f"Epoch {epoch+1}: w = {w:.3f}, loss = {total_loss:.3f}")
```

```python
import numpy as np
import matplotlib.pyplot as plt

# 2D function
def f(w1, w2):
    return (w1 - 3)**2 + (w2 - 2)**2

# Gradients
def grad_f(w1, w2):
    df_dw1 = 2 * (w1 - 3)
    df_dw2 = 2 * (w2 - 2)
    return df_dw1, df_dw2

# Initial point (bad start)
w1, w2 = -4, 6

lr = 0.1  # learning rate
steps = 20

w1_history = []
w2_history = []
loss_history = []

for step in range(steps):
    loss = f(w1, w2)
    g1, g2 = grad_f(w1, w2)
    
    # update rules
    w1 = w1 - lr * g1
    w2 = w2 - lr * g2
    
    w1_history.append(w1)
    w2_history.append(w2)
    loss_history.append(loss)

    print(f"Step {step+1}: w1={w1:.3f}, w2={w2:.3f}, loss={loss:.3f}")

# ===============================
# PLOT CONTOUR + GRADIENT PATH
# ===============================

# create grid
w1_vals = np.linspace(-5, 5, 200)
w2_vals = np.linspace(-2, 8, 200)
W1, W2 = np.meshgrid(w1_vals, w2_vals)
Z = f(W1, W2)

plt.figure(figsize=(7, 6))
plt.contour(W1, W2, Z, levels=40, cmap="jet")

# plot gradient descent steps
plt.plot(w1_history, w2_history, marker="o", color="red", label="Gradient path")
plt.scatter([3], [2], color="black", label="Minimum (3,2)")
plt.title("Gradient Descent Path on 2D Function")
plt.xlabel("w1")
plt.ylabel("w2")
plt.legend()
plt.grid(True)
plt.show()

# Loss evolution
plt.figure(figsize=(6, 4))
plt.plot(loss_history, marker="o")
plt.title("Loss Over Steps")
plt.xlabel("Step")
plt.ylabel("Loss")
plt.grid(True)
plt.show()
```
