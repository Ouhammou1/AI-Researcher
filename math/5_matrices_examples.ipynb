```python
import numpy as np

x = np.array([[5.0],
              [-3.0]])   

def f(x):
    return x.T @ x
    
y = f(x)
# print(y)


def grad_f(x):
    return 2*x

lr  = 0.1
tol = 1e-3

xx = np.linspace(-6, 6, 200)
yy = np.linspace(-6, 6, 200)
X, Y = np.meshgrid(xx, yy)
Z = X**2 + Y**2



for step in range(25):
    loss = f(x)[0,0]
    grad = grad_f(x)
    grad_norm = np.linalg.norm(grad_f(x))
    
    print(
        f"Step {step+1:03d} | "
        f"x = ({x[0,0]:.6f}, {x[1,0]:.6f}) | "
        f"loss = {loss:.6f} | "
        f"||grad|| = {grad_norm:.6f}"
    )

    # ===== DRAW GRAPH FOR THIS STEP =====
    plt.figure(figsize=(6, 5))
    plt.contour(X, Y, Z, levels=20, cmap="jet")

    # current point
    plt.scatter(x[0, 0], x[1, 0], color="red", s=80, label="Current x")

    # minimum
    plt.scatter(0, 0, color="black", s=60, label="Minimum (0,0)")

    plt.title(
        f"Gradient Descent – Step {step+1}\n"
        f"x = ({x[0,0]:.3f}, {x[1,0]:.3f}), "
        f"loss = {loss:.4f}"
    )
    plt.xlabel("x₁")
    plt.ylabel("x₂")
    plt.legend()
    plt.grid(True)
    plt.show()

    # ===== STOP CONDITION =====
    if grad_norm < tol:
        print("✅ Converged: gradient is small")
        break

    # ===== MATRIX UPDATE =====
    x = x - lr * grad
```
