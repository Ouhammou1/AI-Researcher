```python
import torch as t
import numpy as np 
import matplotlib.pyplot as plt 
```

```python
#y = wx + b

inputs = t.tensor([2,3])
weights = t.tensor([0.5, 0.3])
bias = 0.1
```

```python
weights_sum = (weights[0] * inputs[0]) + (weights[1] * inputs[1]) + bias

print(f"weights_sum = {weights_sum:.2f}")
```

```python
#aplay activation function sigmod

activation = 1 / (1 + t.exp(-weights_sum))

print(f"Activation funtion f(x) = {activation}")
```

```python
#Initialize a NEURON
```

```python
class Neuron:

    def __init__(self , num_inputs):
        self.weights = t.randn(num_inputs) * 0.1
        self.bias = t.tensor([0])

    def weights_sum(self ,inputs):
        return  t.dot(self.weights , inputs) + self.bias

    def segmod(self , x):
        return 1 / (1 + t.exp(-x))

        

```

```python
neuron = Neuron(num_inputs=2)

print(f"weights = {neuron.weights}")
print(f"bias    = {neuron.bias}")

neuron.weights = t.tensor([0.5 , 0.3])
neuron.bias    = 0.1

inputs = t.tensor([2.0 ,3.0])

result  = neuron.weights_sum(inputs)
print(f"result = {result}")

output =  neuron.segmod(result)
print(output)



```
